# ADIS - Basic Usage
Welcome to the ADIS library! This notebook demonstrates the fastest way to get started with the ADIS pipeline.

We'll run the full pipeline (Ingestion -> Cleaning -> EDA -> Feature Engineering -> Feature Selection -> Benchmarking -> AI Critic) in just 3 lines of code.

```python
# Step 1: Import the pipeline
from adis import ADISPipeline
```

```python
# For this example, let's create a quick dummy dataset to work with.
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'age': np.random.randint(18, 80, 100),
    'income': np.random.normal(50000, 15000, 100),
    'churn': np.random.choice([0, 1], 100)
})
df.loc[0:5, 'income'] = np.nan  # Inject some missing values
df.to_csv('sample_data.csv', index=False)
print("Created sample_data.csv")
```

```python
# Step 2: Initialize the pipeline
# Tell it which column you want to predict (the target)
pipeline = ADISPipeline(target_column="churn")

# Step 3: Run it!
results = pipeline.run("sample_data.csv")
```

### Exploring the Results
ADIS returns a rich dictionary containing the results of every stage. Let's look at the AI Critic's vulnerability report:

```python
critic = results['critic']
print(f"Is Production Safe: {critic['is_production_safe']}\n")

for vuln in critic['vulnerabilities']:
    print(f"[{vuln['severity'].upper()}] {vuln['issue']}")
    print(f"Reasoning: {vuln['reasoning']}\n")
```

### Saving the Human-Readable Report
You can save the entire pipeline's logic as a Markdown file, a JSON file, and export the fully cleaned CSV.

```python
pipeline.save_report("adis_demo_output")
print("Check the 'adis_demo_output' folder for your reports!")
```
