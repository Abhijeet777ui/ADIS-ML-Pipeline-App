```python
from adis.agent import AutoResearchAgent

# Initialize the agent
agent = AutoResearchAgent(
    filepath="../tests/test_data.csv",
    target_column="target",
    max_iterations=3  # Keep it small for the demo
)
```

### The Optimization Loop

When you run `.optimize()`, the agent will:
1. Read the dataset schema.
2. Propose a new feature engineering script (Python code).
3. Execute that script to transform the data.
4. Run the ADIS pipeline to train a baseline vs. new model.
5. Have the AI Critic evaluate if the new model is safe and actually better.
6. Loop and iterate based on the Critic's feedback.

```python
# Run the autonomous loop
print("Starting autonomous optimization...")
best_results = agent.optimize()

print("\nOptimization Complete!")
```
